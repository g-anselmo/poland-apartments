```python
import pandas as pd

df = pd.read_csv(r'../data/clean/apartments_clean.csv')
df.head()
```

```python
df.loc[df['is_outlier'] == False, ['price_per_m2','city']].groupby('city')['price_per_m2'].median().sort_values(ascending=False)
```

```python
import duckdb
duckdb.sql(""" 
SELECT 
    city,
    MEDIAN(price_per_m2) AS median_city
FROM '../data/clean/apartments_clean.csv'
WHERE is_outlier = False
GROUP BY city
ORDER BY median_city DESC


 """).df()
```

```python
df['building_era'] = pd.cut(df['buildYear'], bins=[0,1945,1990,2015,2030],labels=['pre-war','communist era','modern','new'])

df
```

```python
df.loc[df['is_outlier'] == False,['price_per_m2','building_era']].groupby('building_era')['price_per_m2'].median().sort_values(ascending=False)
```

```python
duckdb.sql("""
SELECT
    CASE 
        WHEN buildYear <= 1945 THEN 'pre-war'
        WHEN buildYear <= 1990 THEN 'communist era'
        WHEN buildYear <= 2015 THEN 'modern'
        WHEN buildYear <= 2030 THEN 'new'
    END as building_era,
    MEDIAN(price_per_m2) AS median_era
FROM '../data/clean/apartments_clean.csv'
WHERE is_outlier = False
GROUP BY building_era
ORDER BY median_era DESC
""").df()
```

```python
qtd = df['buildYear'].isnull().sum()

percent = round(100*(qtd/len(df['buildYear'])),2)
print(percent)
```

```python
near_centrum = df[df['is_outlier'] == False]['centreDistance'].corr(df['price_per_m2'])
near_centrum_warsaw = df[(df['city'] == 'warszawa') & (df['is_outlier'] == False)]['centreDistance'].corr(df['price_per_m2'])
print(near_centrum,near_centrum_warsaw)
```

```python
med_warsaw = df[(df['is_outlier'] == False) & (df['city'] == 'warszawa')].groupby('month')['price_per_m2'].median()
varx  = med_warsaw.shift()

total =  med_warsaw - varx

perc_median = round(((med_warsaw/varx)-1)*100,2)

print(perc_median)
```

```python
total_growth = round((med_warsaw.iloc[-1] / med_warsaw.iloc[0] - 1)*100,2)
total_growth
```

```python
duckdb.sql("""
WITH monthly AS (
    SELECT
        month,
        MEDIAN(price_per_m2) AS median_m2
    FROM '../data/clean/apartments_clean.csv'
    WHERE city = 'warszawa' AND is_outlier = False
    GROUP BY month
)
SELECT
    month,
    median_m2,
    LAG(median_m2) OVER (ORDER BY month) AS prev_median,
    ROUND((median_m2 / LAG(median_m2) OVER (ORDER BY month) - 1) * 100, 2) AS pct_change
FROM monthly
ORDER BY month
""").df()
```
