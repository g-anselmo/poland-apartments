```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
apt_df = pd.read_csv(r'C:\Users\Gustavo Anselmo\Documents\poland-apartments\data\clean\apartments_clean.csv')

waw = apt_df[(apt_df['is_outlier'] == False) & (apt_df['city'] == 'warszawa')]

waw_ml = waw.dropna(subset='buildYear')

x = waw_ml[['buildYear', 'squareMeters', 'centreDistance']]
y = waw_ml['price']

print(x.shape,y.shape)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

```python
model = LinearRegression()

model.fit(x_train, y_train)
model.score(x_test, y_test)
```
